# Lab 01.4 – 4-bit Counter Controlled by Push Buttons

## Objectives
- Combine button inputs and LED outputs.
- Detect button press events using polling.
- Implement a 4-bit up/down counter.
- Apply simple software debouncing.

## Controls
- **BTN0**: increment counter
- **BTN1**: decrement counter
- **BTN2**: reset counter to 0
- **BTN3**: stop the program

The four LEDs display the current counter value in binary.


```python
from pynq import Overlay
import time

ol = Overlay("base.bit")

buttons = ol.btns_gpio.channel1
buttons.setdirection("in")
buttons.setlength(4)

leds = ol.leds_gpio.channel1
leds.setdirection("out")
leds.setlength(4)

print("GPIO initialized.")
```

## Exercise – Up/Down Counter

The program detects a new button press by comparing the current button value with the previous one.
This avoids repeated counting while a button is held down.

A short delay is also used as a simple debounce mechanism.


```python
count = 0
previous = 0

leds.write(count, 0b1111)
print(f"Counter = {count:2d} ({count:04b})")

while True:
    current = buttons.read() & 0xF
    pressed = current & ~previous

    if pressed & 0b0001:          # BTN0
        count = (count + 1) & 0xF

    if pressed & 0b0010:          # BTN1
        count = (count - 1) & 0xF

    if pressed & 0b0100:          # BTN2
        count = 0

    if pressed & 0b1000:          # BTN3
        print("Program stopped.")
        break

    if pressed:
        leds.write(count, 0b1111)
        print(f"Counter = {count:2d} ({count:04b})")
        time.sleep(0.05)

    previous = current
    time.sleep(0.01)

leds.write(0, 0b1111)
```
