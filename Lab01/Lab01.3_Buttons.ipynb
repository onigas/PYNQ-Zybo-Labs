# Lab 01.3 – Reading Push Buttons

## Objectives
- Read the four push buttons through AXI GPIO.
- Identify which button is pressed.
- Use buttons to control the user LEDs.

## Principle
The base overlay contains `btns_gpio`, a 4-bit AXI GPIO input connected to the push buttons.


```python
from pynq import Overlay
import time

ol = Overlay("base.bit")

buttons = ol.btns_gpio.channel1
buttons.setdirection("in")
buttons.setlength(4)

leds = ol.leds_gpio.channel1
leds.setdirection("out")
leds.setlength(4)

print("Button and LED GPIO initialized.")
```

## Exercise 1 – Read the button value


```python
value = buttons.read()
print(f"Button value = {value:04b}")
```

## Exercise 2 – Display button state on LEDs

While a button is pressed, the corresponding LED should turn on.

Stop the cell with **Kernel → Interrupt Kernel**.


```python
while True:
    value = buttons.read()
    leds.write(value, 0b1111)
    time.sleep(0.02)
```
