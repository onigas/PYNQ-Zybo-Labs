# Lab 01.1 – LED Control on Zybo with PYNQ

## Objectives
- Load the PYNQ base overlay.
- Access the AXI GPIO peripheral connected to the four user LEDs.
- Control individual LEDs from Python.
- Generate a simple running-light sequence.

## Hardware
- Digilent Zybo legacy board
- PYNQ 3.0.1
- Ethernet connection between the laptop and Zybo

## Background
The four user LEDs are connected to an AXI GPIO peripheral in the programmable logic (PL).
Python running on the ARM processor accesses this peripheral through the PYNQ API.


```python
from pynq import Overlay
import time

ol = Overlay("base.bit")

leds = ol.leds_gpio.channel1
leds.setdirection("out")
leds.setlength(4)

print("LED GPIO initialized.")
```

## Exercise 1 – Turn LEDs on and off

Each bit controls one LED:

- bit 0 → LD0
- bit 1 → LD1
- bit 2 → LD2
- bit 3 → LD3


```python
# Turn on LD0 only
leds.write(0b0001, 0b1111)
```

```python
# Turn on all four LEDs
leds.write(0b1111, 0b1111)
```

```python
# Turn off all four LEDs
leds.write(0b0000, 0b1111)
```

## Exercise 2 – Running light

Run the next cell and observe LD0 → LD1 → LD2 → LD3.

Stop the cell with **Kernel → Interrupt Kernel**.


```python
while True:
    for i in range(4):
        leds.write(1 << i, 0b1111)
        time.sleep(0.3)
```
