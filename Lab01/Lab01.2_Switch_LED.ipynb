# Lab 01.2 – Reading Switches and Driving LEDs

## Objectives
- Read the four slide switches through AXI GPIO.
- Display the switch value in binary.
- Copy the switch state directly to the four LEDs.

## Principle
The base overlay contains:
- `switches_gpio` – 4-bit input
- `leds_gpio` – 4-bit output


```python
from pynq import Overlay
import time

ol = Overlay("base.bit")

switches = ol.switches_gpio.channel1
switches.setdirection("in")
switches.setlength(4)

leds = ol.leds_gpio.channel1
leds.setdirection("out")
leds.setlength(4)

print("Switch and LED GPIO initialized.")
```

## Exercise 1 – Read the switch value


```python
value = switches.read()
print(f"Switch value = {value:04b}")
```

## Exercise 2 – Copy switches to LEDs

Change the slide switches and observe the LEDs.

Stop the cell with **Kernel → Interrupt Kernel**.


```python
while True:
    value = switches.read()
    leds.write(value, 0b1111)
    time.sleep(0.05)
```
