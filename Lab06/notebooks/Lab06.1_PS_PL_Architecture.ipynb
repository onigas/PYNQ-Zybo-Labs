# Lab06.1 – PS–PL Architecture

**Objective:** identify the ARM PS, programmable logic and AXI-connected IP blocks without reprogramming the FPGA. This exercise uses the **existing** `base.bit` configuration.

**Step 1 – Environment.** Import PYNQ and inspect the board’s installed overlays. The listing confirms which designs are physically available.

```python
from pynq import Overlay, MMIO
import os
import time
import numpy as np
print('PYNQ directory:', os.path.exists('/home/xilinx/pynq/overlays'))
print('Available overlays:', os.listdir('/home/xilinx/pynq/overlays'))
```

**Step 2 – Inspect metadata.** Load the base overlay metadata with `download=False`; list each hardware IP block and its type. This does not program the PL.

```python
base = Overlay('base.bit', download=False)
for name, item in sorted(base.ip_dict.items()):
    print('{:<24} {}'.format(name, item.get('type', 'unknown')))
```

**Step 3 – AXI address map.** Print physical base addresses and address ranges of AXI-mapped IP. AXI peripherals expose memory-mapped registers to the ARM.

```python
for name, item in sorted(base.ip_dict.items()):
    if 'phys_addr' in item:
        print('{:<24} 0x{:08X}, range 0x{:X}'.format(name, item['phys_addr'], item.get('addr_range', 0)))
```

**Step 4 – Concept check.** Confirm which blocks are GPIO and whether a dedicated image or MAC accelerator is present in `base.bit`.

```python
names = list(base.ip_dict)
print('GPIO blocks:', [n for n in names if 'gpio' in n.lower()])
print('Custom MAC available:', any('mac' in n.lower() for n in names))
print('PS–PL concept: ARM controls IP in PL through address-mapped AXI.')
```
