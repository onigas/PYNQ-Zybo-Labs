## PWM Principle
PWM controls the average delivered power by changing the fraction of time a digital output is ON.

Duty cycle:

\[
D = \frac{t_{ON}}{T} \times 100\%
\]

where `T` is the PWM period.

## Note
This experiment uses software PWM. Because Python runs under Linux, the waveform is not precisely timed. The purpose is to understand the PWM principle.


```python
from pynq import Overlay
import time

ol = Overlay("base.bit")

leds = ol.leds_gpio.channel1
leds.setdirection("out")
leds.setlength(4)

print("LED GPIO initialized.")
```

## Exercise 1 – Generate PWM on LD0


```python
def software_pwm(duty_cycle, frequency=100, duration=3):
    duty_cycle = max(0.0, min(100.0, duty_cycle))

    period = 1.0 / frequency
    on_time = period * duty_cycle / 100.0
    off_time = period - on_time

    end_time = time.perf_counter() + duration

    while time.perf_counter() < end_time:
        if on_time > 0:
            leds.write(0b0001, 0b1111)
            time.sleep(on_time)

        if off_time > 0:
            leds.write(0b0000, 0b1111)
            time.sleep(off_time)

    leds.write(0, 0b1111)
```

Run the following cells and compare the apparent brightness.


```python
software_pwm(10, frequency=100, duration=3)
```

```python
software_pwm(50, frequency=100, duration=3)
```

```python
software_pwm(90, frequency=100, duration=3)
```
