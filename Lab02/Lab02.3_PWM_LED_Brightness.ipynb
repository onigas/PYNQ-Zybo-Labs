# Lab 02.3 – LED Brightness Fade Using PWM

## Objectives
- Vary PWM duty cycle automatically.
- Create a fade-in and fade-out effect.
- Observe the nonlinear relationship between PWM duty cycle and perceived brightness.

## Principle
The duty cycle is changed gradually from 0% to 100% and back to 0%.


```python
from pynq import Overlay
import time

ol = Overlay("base.bit")

leds = ol.leds_gpio.channel1
leds.setdirection("out")
leds.setlength(4)

print("LED GPIO initialized.")
```

```python
def pwm_cycle(duty_cycle, frequency=100, cycles=20):
    duty_cycle = max(0.0, min(100.0, duty_cycle))

    period = 1.0 / frequency
    on_time = period * duty_cycle / 100.0
    off_time = period - on_time

    for _ in range(cycles):
        if on_time > 0:
            leds.write(0b0001, 0b1111)
            time.sleep(on_time)

        if off_time > 0:
            leds.write(0b0000, 0b1111)
            time.sleep(off_time)
```

## Exercise 1 – Fade in and fade out

Run the cell below and observe LD0.


```python
for duty in range(0, 101, 10):
    print(f"Duty cycle: {duty}%")
    pwm_cycle(duty, frequency=100, cycles=20)

for duty in range(100, -1, -10):
    print(f"Duty cycle: {duty}%")
    pwm_cycle(duty, frequency=100, cycles=20)

leds.write(0, 0b1111)
```

## Exercise 2 – Repeat the fade continuously

Stop the cell with **Kernel → Interrupt Kernel**.


```python
while True:
    for duty in range(0, 101, 5):
        pwm_cycle(duty, frequency=100, cycles=5)

    for duty in range(100, -1, -5):
        pwm_cycle(duty, frequency=100, cycles=5)
```
