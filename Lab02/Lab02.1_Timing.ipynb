# Lab 02.1 – Timing and Periodic Signals

## Objectives
- Use Python timing functions on PYNQ.
- Generate periodic LED signals.
- Measure elapsed time.
- Observe the difference between requested delay and actual execution time.

## Hardware
- Digilent Zybo legacy board
- PYNQ 3.0.1
- User LEDs connected through the `leds_gpio` AXI GPIO peripheral

## Note
Linux is not a hard real-time operating system. Timing generated in Python is suitable for introductory experiments, but it is not cycle-accurate.


```python
from pynq import Overlay
import time

ol = Overlay("base.bit")

leds = ol.leds_gpio.channel1
leds.setdirection("out")
leds.setlength(4)

leds.write(0, 0b1111)

print("LED GPIO initialized.")
```

## Exercise 1 – Blink LD0

Generate a square-wave-like visual signal on LD0 with a period of approximately 1 second.


```python
for _ in range(10):
    leds.write(0b0001, 0b1111)
    time.sleep(0.5)

    leds.write(0b0000, 0b1111)
    time.sleep(0.5)

leds.write(0, 0b1111)
```

## Exercise 2 – Measure the actual execution time

`time.perf_counter()` provides a high-resolution timer suitable for measuring elapsed time.


```python
t0 = time.perf_counter()

for _ in range(10):
    leds.write(0b0001, 0b1111)
    time.sleep(0.1)
    leds.write(0b0000, 0b1111)
    time.sleep(0.1)

t1 = time.perf_counter()

elapsed = t1 - t0
print(f"Measured execution time: {elapsed:.6f} s")
print("Ideal execution time:    2.000000 s")
print(f"Difference:              {elapsed - 2.0:.6f} s")
```

## Exercise 3 – Change the blinking frequency

Modify `delay` and observe the LED behavior.


```python
delay = 0.05

for _ in range(50):
    leds.write(0b0001, 0b1111)
    time.sleep(delay)
    leds.write(0b0000, 0b1111)
    time.sleep(delay)

leds.write(0, 0b1111)
```
